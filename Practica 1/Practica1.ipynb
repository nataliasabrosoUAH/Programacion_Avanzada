```python
# Ejercicio 2

def is_leap_year(year: int) -> bool:
    """Devuelve True si el año es bisiesto y False si no lo es."""
    return (year % 4 == 0 and year % 100 != 0) or (year % 400 == 0)

def leap_years_between(start: int, end: int) -> list[int]:
    """Devuelve una lista de los años bisiestos entre start y end incluidos."""
    leap_years = []
    
    for year in range(start, end + 1):
        if is_leap_year(year):
            leap_years.append(year)
    return leap_years

def explain_leap_year(year: int) -> str:
    """Indica si el año es bisiesto y explica el motivo."""
    if year % 400 == 0:
        return f"{year} es un año bisiesto: es divisible entre 400."
    elif year % 100 == 0:
        return f"{year} no es un año bisiesto: es divisible entre 100 pero no entre 400."
    elif year % 4 == 0:
        return f"{year} es un año bisiesto: es divisible entre 4 pero no entre 100."
    else:
        return f"{year} no es un año bisiesto: no es divisible entre 4."

def main() -> None:
    """Solicita un año y prueba las funciones del ejercicio."""
    year = int(input("Año: "))
    
    if year < 1582:
        print("El calendario gregoriano se estableció en 1582.")
    else:
        print(explain_leap_year(year))
        
    # Pruebas con los años del enunciado y un caso adicional.
    for test_year in [1900, 2000, 2023, 2024, 2100]:
        print(test_year, is_leap_year(test_year))
        print(explain_leap_year(test_year))
    # Pruebas de los intervalos.
    print(leap_years_between(1896, 1912))
    print(leap_years_between(2020, 2025))
    print(leap_years_between(2021, 2023))
    
main()
```

```python
# Ejercicio 4

def shift_char(char: str, shift: int) -> str:
    """Desplaza una letra inglesa y deja los demás caracteres sin cambios."""
    if 'a' <= char <= 'z':
        base = ord('a')
    elif 'A' <= char <= 'Z':
        base = ord('A')
    else:
        return char
    
    position = ord(char) - base
    new_position = (position + shift) % 26

    return chr(base + new_position)

def encrypt(text: str, shift: int) -> str:
    """Cifra un texto desplazando cada letra las posiciones indicadas."""
    encrypted_text = ""
    
    for char in text:
        encrypted_text += shift_char(char, shift)
        
    return encrypted_text

def decrypt(text: str, shift: int) -> str:
    """Descifra un texto utilizando el desplazamiento contrario."""
    return encrypt(text, -shift)

def brute_force(cipher_text: str) -> list[str]:
    """Devuelve los 26 posibles descifrados del texto."""
    results = []

    for shift in range(26):
        results.append(decrypt(cipher_text, shift))

    return results

def main() -> None:
    """Prueba las funciones del cifrado César."""
    # Desplazamiento de una letra.
    print("Letra b desplazada 3 posiciones:", shift_char("b", 3))

    # Cifrado y descifrado de un texto.
    print("Texto cifrado:", encrypt("Hola, Mundo!", 3))
    print("Texto descifrado:", decrypt("Krod, Pxqgr!", 3))

    # Vuelta al principio del alfabeto y desplazamiento negativo.
    print("xyz con desplazamiento 3:", encrypt("xyz", 3))
    print("abc con desplazamiento -3:", encrypt("abc", -3))

    # Desplazamiento mayor que 26 y caracteres que no cambian.
    print("abc con desplazamiento 29:", encrypt("abc", 29))
    print("Números y signos:", encrypt("123!?", 3))

    # Los 26 posibles descifrados.
    print("Posibles descifrados:")
    candidates = brute_force("Krod, Pxqgr!")

    for shift in range(26):
        print(shift, candidates[shift])


main()
```
