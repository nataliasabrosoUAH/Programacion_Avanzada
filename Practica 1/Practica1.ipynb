```python
# Ejercicio 2

def is_leap_year(year: int) -> bool:
    """Devuelve True si el año es bisiesto y False si no lo es."""
    return (year % 4 == 0 and year % 100 != 0) or (year % 400 == 0)
print(is_leap_year(2024))
print(is_leap_year(1900))
print(is_leap_year(2000))
print(is_leap_year(2023))

def leap_years_between(start: int, end: int) -> list[int]:
    """Devuelve una lista de los años bisiestos entre start y end incluidos."""
    leap_years = []
    
    for year in range(start, end + 1):
        if is_leap_year(year):
            leap_years.append(year)
    return leap_years
print(leap_years_between(1896, 1912))
print(leap_years_between(2020, 2025))

def explain_leap_year(year: int) -> str:
    """Indica si el año es bisiesto y explica el motivo."""
    if year % 400 == 0:
        return f"{year} es un año bisiesto: es divisible entre 400."
    elif year % 100 == 0:
        return f"{year} no es un año bisiesto: es divisible entre 100 pero no entre 400."
    elif year % 4 == 0:
        return f"{year} es un año bisiesto: es divisible entre 4 pero no entre 100."
    else:
        return f"{year} no es un año bisiesto: no es divisible entre 4."
    
print(explain_leap_year(1900))
print(explain_leap_year(2000))
print(explain_leap_year(2023))
print(explain_leap_year(2024))

def main() -> None:
    """Solicita un año y prueba las funciones del ejercicio."""
    year = int(input("Año: "))
    
    if year < 1582:
        print("El calendario gregoriano se estableció en 1582.")
    else:
        print(explain_leap_year(year))
        
    # Pruebas con los años del enunciado y un caso adicional.
    for test_year in [1900, 2000, 2023, 2024, 2100]:
        print(test_year, is_leap_year(test_year))
        print(explain_leap_year(test_year))
    # Pruebas de los intervalos.
    print(leap_years_between(1896, 1912))
    print(leap_years_between(2020, 2025))
    print(leap_years_between(2021, 2023))
    
main()
```
