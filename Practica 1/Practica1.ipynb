```python
#Ejercicio 1
def is_integer(text: str) -> bool:
    """Comprueba si el texto introducido representa un n├║mero entero."""
    try:
        int(text)
        return True
    except ValueError:
        return False

def seconds_to_dhms(total_seconds: int) -> tuple[int, int, int, int]:
    """Convierte una cantidad de segundos en d├¡as, horas, minutos y segundos."""
    days = total_seconds // 86400
    hours = (total_seconds % 86400) // 3600
    minutes = (total_seconds % 3600) // 60
    seconds = total_seconds % 60
    return days, hours, minutes, seconds
def main() -> None:
    """Solicita una cantidad de segundos y muestra su conversi├│n."""
    while True:
        text= input("Introduzca los segundos: ")
        if is_integer(text):
            total_seconds = int(text)
            if total_seconds >= 0:
                break
        print("Entrada no valida. Introduzca un entero no negativo. ")
    days, hours, minutes, seconds =seconds_to_dhms(total_seconds)
    print(f"{total_seconds} segundos son {days} d, {hours} h, {minutes} min y {seconds} s.")
main()
```

```python
# Ejercicio 2

def is_leap_year(year: int) -> bool:
    """Devuelve True si el año es bisiesto y False si no lo es."""
    return (year % 4 == 0 and year % 100 != 0) or (year % 400 == 0)

def leap_years_between(start: int, end: int) -> list[int]:
    """Devuelve una lista de los años bisiestos entre start y end incluidos."""
    leap_years = []
    
    for year in range(start, end + 1):
        if is_leap_year(year):
            leap_years.append(year)
    return leap_years

def explain_leap_year(year: int) -> str:
    """Indica si el año es bisiesto y explica el motivo."""
    if year % 400 == 0:
        return f"{year} es un año bisiesto: es divisible entre 400."
    elif year % 100 == 0:
        return f"{year} no es un año bisiesto: es divisible entre 100 pero no entre 400."
    elif year % 4 == 0:
        return f"{year} es un año bisiesto: es divisible entre 4 pero no entre 100."
    else:
        return f"{year} no es un año bisiesto: no es divisible entre 4."

def main() -> None:
    """Solicita un año y prueba las funciones del ejercicio."""
    year = int(input("Año: "))
    
    if year < 1582:
        print("El calendario gregoriano se estableció en 1582.")
    else:
        print(explain_leap_year(year))
        
    # Pruebas con los años del enunciado y un caso adicional.
    for test_year in [1900, 2000, 2023, 2024, 2100]:
        print(test_year, is_leap_year(test_year))
        print(explain_leap_year(test_year))
    # Pruebas de los intervalos.
    print(leap_years_between(1896, 1912))
    print(leap_years_between(2020, 2025))
    print(leap_years_between(2021, 2023))
    
main()
```

```python
# Ejercicio 4

def shift_char(char: str, shift: int) -> str:
    """Desplaza una letra inglesa y deja los demás caracteres sin cambios."""
    if 'a' <= char <= 'z':
        base = ord('a')
    elif 'A' <= char <= 'Z':
        base = ord('A')
    else:
        return char
    
    position = ord(char) - base
    new_position = (position + shift) % 26

    return chr(base + new_position)

def encrypt(text: str, shift: int) -> str:
    """Cifra un texto desplazando cada letra las posiciones indicadas."""
    encrypted_text = ""
    
    for char in text:
        encrypted_text += shift_char(char, shift)
        
    return encrypted_text

def decrypt(text: str, shift: int) -> str:
    """Descifra un texto utilizando el desplazamiento contrario."""
    return encrypt(text, -shift)

def brute_force(cipher_text: str) -> list[str]:
    """Devuelve los 26 posibles descifrados del texto."""
    results = []

    for shift in range(26):
        results.append(decrypt(cipher_text, shift))

    return results

def main() -> None:
    """Prueba las funciones del cifrado César."""
    # Desplazamiento de una letra.
    print("Letra b desplazada 3 posiciones:", shift_char("b", 3))

    # Cifrado y descifrado de un texto.
    print("Texto cifrado:", encrypt("Hola, Mundo!", 3))
    print("Texto descifrado:", decrypt("Krod, Pxqgr!", 3))

    # Vuelta al principio del alfabeto y desplazamiento negativo.
    print("xyz con desplazamiento 3:", encrypt("xyz", 3))
    print("abc con desplazamiento -3:", encrypt("abc", -3))

    # Desplazamiento mayor que 26 y caracteres que no cambian.
    print("abc con desplazamiento 29:", encrypt("abc", 29))
    print("Números y signos:", encrypt("123!?", 3))

    # Los 26 posibles descifrados.
    print("Posibles descifrados:")
    candidates = brute_force("Krod, Pxqgr!")

    for shift in range(26):
        print(shift, candidates[shift])


main()
```

```python
def normalize(text: str) -> str:
    """Convierte un texto a minúsculas y elimina tildes y caracteres no alfanuméricos."""
    text = text.lower()

    text = text.replace("á", "a")
    text = text.replace("é", "e")
    text = text.replace("í", "i")
    text = text.replace("ó", "o")
    text = text.replace("ú", "u")
    text = text.replace("ü", "u")

    result = ""

    for char in text:
        if char.isalnum():
            result += char

    return result


def is_palindrome(text: str) -> bool:
    """Comprueba si un texto es un palíndromo."""
    normalized_text = normalize(text)
    reversed_text = normalized_text[::-1]

    return normalized_text == reversed_text


def find_palindromes(sentences: list[str]) -> list[str]:
    """Devuelve una lista con las frases que son palíndromos."""
    palindromes = []

    for sentence in sentences:
        if is_palindrome(sentence):
            palindromes.append(sentence)

    return palindromes


def palindrome_words(text: str) -> list[str]:
    """Devuelve las palabras palíndromas de al menos tres letras."""
    words = text.split()
    result = []

    for word in words:
        if len(word) >= 3:
            if is_palindrome(word):
                result.append(word)

    return result


def main() -> None:
    """Prueba todas las funciones del ejercicio 5."""

    # Prueba de normalize
    print(normalize("¡Ánita lava la tina!"))

    # Pruebas de is_palindrome
    print(is_palindrome("Anita lava la tina"))
    print(is_palindrome("Hola mundo"))

    # Prueba de find_palindromes
    sentences = [
        "Anita lava la tina",
        "Dábale arroz a la zorra el abad",
        "Esto no es un palíndromo",
        "¿Acaso hubo búhos acá?"
    ]

    print(find_palindromes(sentences))

    # Prueba de palindrome_words
    print(palindrome_words("Ana vio un oso en el ojo de Anita"))

    # Pruebas adicionales
    print(is_palindrome("Reconocer"))
    print(palindrome_words("Somos seres"))


main()


```

```python
# Ejercicio 6

def is_prime(n: int) -> bool:
    """Devuelve True si n es primo y False si no lo es."""
    if n < 2:
        return False

    for divisor in range(2, n):
        if n % divisor == 0:
            return False

    return True

def prime_list(limit: int) -> list[int]:
    """Devuelve los números primos hasta limit, incluido."""
    primes = []  # crea una lista vacia

    for number in range(2, limit + 1): #recorre los numeros desde 2 hasta el límite. Sumamos 1 porque range no incluye su extremo final.
        if is_prime(number):
            primes.append(number)

    return primes

def check_palindrome(primes: list[int]) -> list[int]:
    """Devuelve los números palíndromos de una lista de primos."""
    palindromes = []

    for prime in primes:
        text = str(prime) #convierte el numero en texto 
        if text == text[::-1]: #obtiene el texto al reves y lo compara con el original
            palindromes.append(prime)

    return palindromes

def categorize_prime(prime: int) -> str:
    """Clasifica un número primo como pequeño, mediano o grande."""
    if prime < 10:
        return "pequeño"
    elif prime < 100:
        return "mediano"
    else:
        return "grande"
    
def main() -> None:
    """Prueba las funciones de análisis de números primos."""
    # Comprobar números primos y no primos.
    for number in [-3, 0, 1, 2, 7, 9]:
        print(number, "¿es primo?", is_prime(number))

    # Generar listas con diferentes límites.
    for limit in [1, 2, 7, 20]:
        print("Primos hasta", limit, ":", prime_list(limit))

    # Buscar palíndromos entre los primos hasta 150.
    primes = prime_list(150)
    print("Primos palíndromos hasta 150:", check_palindrome(primes))

    # Probar las tres categorías.
    for prime in [7, 13, 97, 101]:
        print(prime, "es", categorize_prime(prime))


main()
```

```python
from typing import List, Tuple, Optional

def remove_duplicates(numbers: List[int]) -> List[int]:
    """
    Devuelve una nueva lista sin elementos repetidos,
    conservando el orden de la primera aparición de cada valor.
    """
    result = []
    for num in numbers:
        if num not in result:
            result.append(num)
    return result

def rotate_left(numbers: List[int], k: int) -> List[int]:
    """
    Devuelve una nueva lista con los elementos rotados k posiciones hacia la izquierda.
    Funciona si k es mayor que la longitud, negativo, cero o si la lista está vacía.
    """
    if not numbers:
        return []
    
    n = len(numbers)
    # Ajustar k en caso de que sea mayor que la longitud o negativo
    effective_k = k % n
    
    # Realizamos la rotación mediante slicing sin modificar la original
    return numbers[effective_k:] + numbers[:effective_k]

def split_even_odd(numbers: List[int]) -> Tuple[List[int], List[int]]:
    """
    Devuelve una tupla con dos listas: los números pares y los impares, en su orden original.
    """
    evens = []
    odds = []
    for num in numbers:
        if num % 2 == 0:
            evens.append(num)
        else:
            odds.append(num)
    return (evens, odds)

def merge_sorted(a: List[int], b: List[int]) -> List[int]:
    """
    Recibe dos listas ya ordenadas de menor a mayor y devuelve una única lista ordenada,
    recorriendo ambas a la vez con dos índices.
    """
    result = []
    i = 0
    j = 0
    
    while i < len(a) and j < len(b):
        if a[i] <= b[j]:
            result.append(a[i])
            i += 1
        else:
            result.append(b[j])
            j += 1
            
    # Añadir los elementos restantes de la lista 'a' si los hay
    while i < len(a):
        result.append(a[i])
        i += 1
        
    # Añadir los elementos restantes de la lista 'b' si los hay
    while j < len(b):
        result.append(b[j])
        j += 1
        
    return result

def second_largest(numbers: List[int]) -> Optional[int]:
    """
    Devuelve el segundo valor más grande (distinto del máximo) recorriendo la lista una sola vez.
    Si no existe (lista vacía, un solo elemento o todos iguales), devuelve None.
    """
    if len(numbers) < 2:
        return None
    
    max_val = None
    second_max = None
    
    for num in numbers:
        if max_val is None or num > max_val:
            second_max = max_val
            max_val = num
        elif num < max_val:
            if second_max is None or num > second_max:
                second_max = num
                
    return second_max

def main() -> None:
    """Función principal para probar los ejercicios con los ejemplos y casos adicionales."""
    
    # Pruebas de remove_duplicates
    print("--- Pruebas de remove_duplicates ---")
    orig1 = [3, 1, 3, 2, 1]
    print(f"Original: {orig1}")
    print(f"Sin duplicados: {remove_duplicates(orig1)}")
    
    # Pruebas de rotate_left
    print("\n--- Pruebas de rotate_left ---")
    orig2 = [1, 2, 3, 4, 5]
    print(f"Original: {orig2}")
    print(f"Rotar 2: {rotate_left(orig2, 2)}")
    print(f"Rotar 7: {rotate_left(orig2, 7)}")
    print(f"Rotar -1: {rotate_left(orig2, -1)}")
    print(f"Rotar 0: {rotate_left(orig2, 0)}")
    
    # Pruebas de split_even_odd
    print("\n--- Pruebas de split_even_odd ---")
    orig3 = [1, 2, 3, 4, 5, 6]
    print(f"Original: {orig3}")
    print(f"Pares e impares: {split_even_odd(orig3)}")
    
    # Pruebas de merge_sorted
    print("\n--- Pruebas de merge_sorted ---")
    list_a = [1, 4, 9]
    list_b = [2, 3, 10, 11]
    print(f"Listas a combinar: {list_a} y {list_b}")
    print(f"Resultado fusionado: {merge_sorted(list_a, list_b)}")
    
    # Pruebas de second_largest
    print("\n--- Pruebas de second_largest ---")
    print(f"Segundo mayor de [4, 9, 2, 9, 7]: {second_largest([4, 9, 2, 9, 7])}")
    print(f"Segundo mayor de [5, 5]: {second_largest([5, 5])}")
    print(f"Segundo mayor de [10]: {second_largest([10])}")

# Ejecutar el main si se ejecuta el script
if __name__ == "__main__":
    main()
```

```python
# Ejercicio 8

def sum_numbers(numbers: list[int]) -> int:
    """Devuelve la suma de todos los n├║meros de la lista."""
    total = 0
    
    for number in numbers:
        total += number
    return total

def sum_positive_negative(numbers: list[int]) -> tuple[int, int]:
    """Devuelve la suma de los n├║meros positivos y negativos de la lista."""
    positive_sum = 0
    negative_sum = 0
    
    for number in numbers:
        if number > 0:
            positive_sum += number
        elif number < 0:
            negative_sum += number
            
    return positive_sum, negative_sum

def find_max_min(numbers: list[int]) -> tuple[int, int] | None:
    """Devuelve el m├íximo y el m├¡nimo de la lista, o None si la lista esta vac├¡a."""
    if not numbers:
        return None
    
    maximum = max(numbers)
    minimum = min(numbers)
    
    return maximum, minimum

def all_unique(numbers: list[int]) -> bool:
    """Devuelve True si no hay n├║meros repetidos y False si los hay."""
    seen = []
    
    for number in numbers:
        if number in seen:
            return False
        seen.append(number)
        
    return True

def compare_positive_negative(numbers: list[int]) -> str:
    """Indica si hay m├ís positivos, m├ís negativos o la misma cantidad."""
    positive_count = 0
    negative_count = 0
    
    for number in numbers:
        if number > 0:
            positive_count += 1
        elif number < 0:
            negative_count += 1
            
    if positive_count == negative_count:
        return "Hay tantos positivos como negativos."
    
    return(
        "Hay m├ís n├║meros positivos."
        if positive_count > negative_count
        else "Hay m├ís n├║meros negativos."
    )
    
def main() -> None:
    """Lee una lista de enteros y permite analizarlos mediante un men├║."""
    while True:
        text = input(
            "Introduce enteros separados por espacios"
            "(Enter para una lista vac├¡a): "
        )
        
        try:
            numbers = []
            for value in text.split():
                numbers.append(int(value))
            break
        except ValueError:
            print("Debes introducir solo n├║meros enteros separados por espacios.")
    while True:
        print("\nLista actual:", numbers)
        print("1. Sumar todos los n├║meros")
        print("2. Sumar positivos y negativos por separado")
        print("3. Mostrar m├íximo y m├¡nimo")
        print("4. Comprobar si todos los n├║meros son ├║nicos")
        print("5. Comparar la cantidad de positivos y negativos")
        print("0. Salir")
        
        option = input("Elige una opci├│n: ").strip()
        
        if option == "1":
            print("Suma total:", sum_numbers(numbers))
            
        elif option == "2":
            positive_sum, negative_sum = sum_positive_negative(numbers)
            print("Suma de positivos:", positive_sum)
            print("Suma de negativos:", negative_sum)
            
        elif option == "3":
            result = find_max_min(numbers)
            
            if result is None:
                print("La lista est├í vac├¡a: no tiene m├íximo ni m├¡nimo.")
            else:
                maximum, minimum = result
                print("Maximo:", maximum)
                print("M├¡nimo:", minimum)
                
        elif option == "4":
            if all_unique(numbers):
                print("Todos los n├║meros son unicos.")
            else:
                print("Hay n├║meros repetidos.")
                
        elif option == "5":
            print(compare_positive_negative(numbers))
            
        elif option == "0":
            print("Programa Terminado.")
            break
        else:
            print("Opci├│n no v├ílida. Elige un n├║mero del 0 al 5.")
main()
```

```python
from typing import List, Union

def format_monomial(coef: Union[int, float], exp: int) -> str:
    """
    Devuelve el monomio como cadena con su signo explícito, aplicando las reglas:
    - Si el coeficiente es 0, devuelve una cadena vacía.
    - Si el coeficiente es 1 o -1 y el exp > 0, no se escribe el 1 (solo el signo).
    - Si el exp es 1, no se escribe el exponente.
    - Si el exp es 0, no se escribe la 'x', solo el coeficiente.
    """
    if coef == 0.0 or coef == 0:
        return ""
    
    # Determinar el signo explícito
    if coef > 0:
        sign = "+"
    else:
        sign = "-"
        
    abs_coef = abs(coef)
    
    # Manejar el formato del coeficiente y la x según el exponente
    if exp == 0:
        # Si es un número entero, mostrarlo sin decimales innecesarios si es .0
        coef_str = str(int(abs_coef)) if isinstance(abs_coef, float) and abs_coef.is_integer() else str(abs_coef)
        return f"{sign}{coef_str}"
    
    # Para exponente > 0
    if abs_coef == 1.0:
        coef_str = ""
    else:
        coef_str = str(int(abs_coef)) if isinstance(abs_coef, float) and abs_coef.is_integer() else str(abs_coef)
        
    if exp == 1:
        return f"{sign}{coef_str}x"
    else:
        return f"{sign}{coef_str}x**{exp}"

def format_polynomial(coefs: List[Union[int, float]]) -> str:
    """
    Devuelve el polinomio completo con los términos ordenados de mayor a menor exponente,
    omitiendo los de coeficiente 0.
    """
    if not coefs or all(c == 0 for c in coefs):
        return "0"
    
    terms = []
    n = len(coefs)
    
    # Recorrer de mayor a menor exponente
    for i in range(n - 1, -1, -1):
        coef = coefs[i]
        monomial = format_monomial(coef, i)
        if monomial:
            terms.append(monomial)
            
    if not terms:
        return "0"
    
    # Unir los términos respetando las reglas de los signos
    result = terms[0]
    # Si el primer término empieza por '+', se lo quitamos
    if result.startswith("+"):
        result = result[1:]
        
    for term in terms[1:]:
        if term.startswith("+"):
            result += f" + {term[1:]}"
        elif term.startswith("-"):
            result += f" - {term[1:]}"
            
    return result

def evaluate(coefs: List[Union[int, float]], x: Union[int, float]) -> Union[int, float]:
    """
    Devuelve el valor del polinomio para un valor x dado (0 si la lista está vacía).
    """
    if not coefs:
        return 0.0
    
    result = 0.0
    for i, coef in enumerate(coefs):
        result += coef * (x ** i)
        
    return result

def derivative(coefs: List[Union[int, float]]) -> List[Union[int, float]]:
    """
    Devuelve la lista de coeficientes del polinomio derivado.
    Si el polinomio es constante o está vacío, devuelve una lista vacía.
    """
    if len(coefs) <= 1:
        return []
    
    deriv_coefs = []
    for i in range(1, len(coefs)):
        deriv_coefs.append(coefs[i] * i)
        
    return deriv_coefs

def main() -> None:
    """Función principal para probar las funciones del Ejercicio 9."""
    
    print("--- Pruebas de format_monomial ---")
    print(f"format_monomial(2.0, 3) -> '{format_monomial(2.0, 3)}'")
    print(f"format_monomial(1.0, 2) -> '{format_monomial(1.0, 2)}'")
    print(f"format_monomial(-1.0, 1) -> '{format_monomial(-1.0, 1)}'")
    print(f"format_monomial(4.0, 0) -> '{format_monomial(4.0, 0)}'")
    print(f"format_monomial(0.0, 5) -> '{format_monomial(0.0, 5)}'")
    
    print("\n--- Pruebas de format_polynomial ---")
    poly1 = [4.0, 1.0, 0.0, 2.0]
    poly2 = [0.0, -1.0, 3.0]
    poly3 = [-2.5, 0.0, -1.0]
    poly4 = [0.0, 0.0]
    
    print(f"format_polynomial({poly1}) -> '{format_polynomial(poly1)}'")
    print(f"format_polynomial({poly2}) -> '{format_polynomial(poly2)}'")
    print(f"format_polynomial({poly3}) -> '{format_polynomial(poly3)}'")
    print(f"format_polynomial({poly4}) -> '{format_polynomial(poly4)}'")
    
    print("\n--- Pruebas de evaluate ---")
    print(f"evaluate({poly1}, 2) -> {evaluate(poly1, 2)}")
    
    print("\n--- Pruebas de derivative ---")
    print(f"derivative({poly1}) -> {derivative(poly1)}")

if __name__ == "__main__":
    main()
```

```python

# Ejercicio 10

def strings_starting_with(strings: list[str], letter: str) -> list[str]:
    """Devuelve las cadenas que comienzan por la letra indicada."""
    result = []

    for string in strings:
        if string.startswith(letter):
            result.append(string)

    return result


def count_substring(strings: list[str], substring: str) -> int:
    """Cuenta cuántas cadenas contienen la subcadena indicada."""
    count = 0

    for string in strings:
        if substring in string:
            count += 1

    return count


def find_longest_shortest(strings: list[str]) -> tuple[str, str] | None:
    """Devuelve la cadena más larga y la más corta, o None si la lista está vacía."""
    if not strings:
        return None

    longest = strings[0]
    shortest = strings[0]

    for string in strings:
        if len(string) > len(longest):
            longest = string
        if len(string) < len(shortest):
            shortest = string

    return longest, shortest


def same_object(strings: list[str], index1: int, index2: int) -> bool:
    """Comprueba si dos cadenas de la lista son el mismo objeto."""
    return strings[index1] is strings[index2]


def check_length(strings: list[str]) -> None:
    """Comprueba con for...else si alguna cadena tiene más de 10 caracteres."""
    for string in strings:
        if len(string) > 10:
            print("Hay al menos una cadena con más de 10 caracteres.")
            break
    else:
        print("Ninguna cadena tiene más de 10 caracteres.")


def main() -> None:
    """Lee una lista de cadenas y permite analizarlas mediante un menú."""
    text = input("Introduce cadenas separadas por comas (Enter para una lista vacía): ")

    strings = []

    if text.strip():
        for string in text.split(","):
            strings.append(string.strip())

    while True:
        print("\nLista actual:", strings)
        print("1. Mostrar cadenas que comienzan por una letra")
        print("2. Contar cadenas que contienen una subcadena")
        print("3. Mostrar la cadena más larga y la más corta")
        print("4. Comprobar si dos cadenas son el mismo objeto")
        print("5. Comprobar si alguna cadena tiene más de 10 caracteres")
        print("0. Salir")

        option = input("Elige una opción: ").strip()

        if option == "1":
            letter = input("Introduce una letra: ")

            if len(letter) != 1 or not letter.isalpha():
                print("Debes introducir una sola letra.")
            else:
                result = strings_starting_with(strings, letter)
                print("Cadenas que comienzan por", letter, ":", result)

        elif option == "2":
            substring = input("Introduce una subcadena: ")
            print("Cantidad de cadenas:", count_substring(strings, substring))

        elif option == "3":
            result = find_longest_shortest(strings)

            if result is None:
                print("La lista está vacía: no tiene cadena más larga ni más corta.")
            else:
                longest, shortest = result
                print("Cadena más larga:", longest)
                print("Cadena más corta:", shortest)

        elif option == "4":
            if len(strings) < 2:
                print("Se necesitan al menos dos cadenas en la lista.")
            else:
                try:
                    index1 = int(input("Introduce la posición de la primera cadena (desde 0): "))
                    index2 = int(input("Introduce la posición de la segunda cadena (desde 0): "))

                    if 0 <= index1 < len(strings) and 0 <= index2 < len(strings):
                        if same_object(strings, index1, index2):
                            print("Las dos cadenas son el mismo objeto.")
                        else:
                            print("Las dos cadenas no son el mismo objeto.")
                    else:
                        print("Las posiciones introducidas no son válidas.")

                except ValueError:
                    print("Debes introducir números enteros.")

        elif option == "5":
            check_length(strings)

        elif option == "0":
            print("Programa terminado.")
            break

        else:
            print("Opción no válida. Elige un número del 0 al 5.")


main()
```

```python
import re
from datetime import datetime

def process_medical_report(report: str) -> str:
    """
    Procesa un informe médico simulado utilizando expresiones regulares para:
    - Calcular la edad a partir de la fecha de nacimiento y la fecha de consulta.
    - Sustituir las demás fechas por 'FECHA'.
    - Anonimizar el nombre del paciente manteniendo la etiqueta.
    - Anonimizar el nombre del médico manteniendo el formato de la línea.
    """
    # Expresión regular para detectar fechas en formato dd/mm/aaaa (con 1 o 2 cifras en día/mes y 4 en año)[cite: 10]
    date_regex = re.compile(r'\b\d{1,2}/\d{1,2}/\d{4}\b')
    
    # 1. Extraer la fecha de la consulta (que se encuentra en la misma línea que el médico)
    consultation_date_str = None
    dates_found = date_regex.findall(report)
    if len(dates_found) > 1:
        # Por lo general, la última fecha del informe (o la del médico) es la de la consulta
        consultation_date_str = dates_found[-1]

    # 2. Procesar la Fecha de nacimiento antes que el resto de fechas[cite: 10]
    birth_regex = re.compile(r'Fecha de nacimiento:\s*(\d{1,2}/\d{1,2}/\d{4})')
    match_birth = birth_regex.search(report)
    
    if match_birth and consultation_date_str:
        birth_date_str = match_birth.group(1)
        birth_date = datetime.strptime(birth_date_str, '%d/%m/%Y')
        consultation_date = datetime.strptime(consultation_date_str, '%d/%m/%Y')
        
        # Calcular la edad del paciente en la fecha de la consulta[cite: 10]
        age = consultation_date.year - birth_date.year
        if (consultation_date.month, consultation_date.day) < (birth_date.month, birth_date.day):
            age -= 1
            
        report = birth_regex.sub(f'Edad: {age}', report)

    # 3. Sustituir el resto de fechas en formato dd/mm/aaaa por la palabra "FECHA"[cite: 10]
    report = date_regex.sub('FECHA', report)

    # 4. Sustituir el nombre del paciente por "PACIENTE", conservando la etiqueta "Nombre:"[cite: 10]
    patient_regex = re.compile(r'Nombre:\s*[^\n]+')
    report = patient_regex.sub('Nombre:\nPACIENTE', report)

    # 5. Sustituir el nombre del médico (Dr./Dra. y lo que sigue hasta el final de la línea) por "MEDICO"[cite: 10]
    doctor_regex = re.compile(r'(Dr\.|Dra\.)\s+[^\n]+')
    report = doctor_regex.sub('MEDICO', report)

    return report

def main() -> None:
    """Función principal para probar el procesamiento del informe médico."""
    informe = """Informe clínico de Urgencias
Nombre: Juan Pérez López
Género: Hombre
Fecha de nacimiento: 12/05/1980
Motivo de consulta: dolor abdominal
12/05/2023 Dr. Ramírez
Tratamiento: reposo y analgésicos"""

    print("--- Informe Original ---")
    print(informe)
    
    print("\n--- Resultado Procesado ---")
    resultado = process_medical_report(informe)
    print(resultado)

if __name__ == "__main__":
    main()
```

```python

# Ejercicio 12

def is_anagram(word1: str, word2: str) -> bool:
    """Comprueba si dos palabras son anagramas, ignorando mayúsculas y espacios."""
    word1 = word1.lower().replace(" ", "")
    word2 = word2.lower().replace(" ", "")

    return sorted(word1) == sorted(word2)


def group_anagrams(words: list[str]) -> list[list[str]]:
    """Agrupa las palabras que son anagramas, conservando su orden."""
    groups = []

    for word in words:
        found = False

        for group in groups:
            if is_anagram(word, group[0]):
                group.append(word)
                found = True
                break

        if not found:
            groups.append([word])
            
    return groups


def largest_group(groups: list[list[str]]) -> list[str]:
    """Devuelve el grupo con más palabras o una lista vacía si no hay grupos."""
    if not groups:
        return []

    largest = groups[0]

    for group in groups:
        if len(group) > len(largest):
            largest = group

    return largest


def find_anagrams(word: str, words: list[str]) -> list[str]:
    """Busca los anagramas de una palabra, excluyendo las palabras iguales."""
    result = []

    for current_word in words:
        if current_word.lower() != word.lower():
            if is_anagram(word, current_word):
                result.append(current_word)

    return result


def main() -> None:
    """Prueba todas las funciones del ejercicio 12."""

    words = ["roma", "amor", "pelo", "mora", "casa", "lope", "ramo"]

    # Pruebas de is_anagram
    print(is_anagram("Roma", "amor"))
    print(is_anagram("casa", "caso"))
    print(is_anagram("dormitory", "dirty room"))

    # Prueba de group_anagrams
    groups = group_anagrams(words)
    print("Grupos de anagramas:", groups)

    # Prueba de largest_group
    print("Grupo más grande:", largest_group(groups))

    # Pruebas de find_anagrams
    print("Anagramas de amor:", find_anagrams("amor", words))
    print("Anagramas con repetidos:", find_anagrams("amor", ["amor", "roma", "AMOR", "roma"]))
    print("Anagramas de sol:", find_anagrams("sol", words))

    # Pruebas adicionales
    print("Lista vacía:", group_anagrams([]))
    print("Grupo más grande de lista vacía:", largest_group([]))
    print("Anagramas de listen:", find_anagrams("listen", ["silent", "enlist", "hello"]))


main()
```
